## Vilken Python kör vi?

Modulen `sys` följer med Python och berättar saker om den Python som kör vår kod. Med `import` gör vi modulen tillgänglig i notebooken.

````python
import sys

print(sys.version)
````

Projektet kräver Python 3.12 eller senare. Raden nedan stoppar med ett fel om versionen är för gammal. Om inget händer är allt som det ska.

````python
assert sys.version_info >= (3, 12), "Python 3.12 eller senare behövs"
````

## Vilka paket har vi?

I serien använder vi några paket utöver det som följer med Python.

- `pandas` för tabeller med data
- `httpx` för att hämta data från webben
- `duckdb` och `pyarrow` för att lagra och fråga data
- `matplotlib` för diagram
- `jupyterlab` för själva notebookmiljön

Funktionen `version` talar om vilken version av ett paket som är installerad. Vi går igenom paketen ett i taget med en `for`-loop och skriver ut namn och version.

````python
from importlib.metadata import version

paket = ["pandas", "httpx", "duckdb", "pyarrow", "matplotlib", "jupyterlab"]

for namn in paket:
    print(namn, version(namn))
````

## Vår första tabell

Samma uppgifter blir tydligare som en tabell. I pandas heter en tabell `DataFrame`. Vi brukar importera pandas under det korta namnet `pd`.

Här bygger vi en tabell med en rad per paket. Varje rad är en ordbok, en `dict`, där nycklarna blir kolumnnamn.

````python
import pandas as pd

rader = [{"paket": namn, "version": version(namn)} for namn in paket]
versioner = pd.DataFrame(rader)
versioner
````

Raden `rader = [...]` är en kortare skrivform för samma loop som ovan. Den bygger en lista med en ordbok per paket.

En tabell i pandas har alltid ett antal rader och kolumner. Det ser vi med `shape`.

````python
versioner.shape
````

````python
from pathlib import Path

Path.cwd()
````
